```python
# Phần 1.1
trang_thai_don_hang = ("Da giao", "Dang giao", "Huy don", "Da hoan tra")
khu_vuc = ["HN", "HN1", "HN2", "HN3"]
kenh_ban = ["TIKTOK", "SHOPPE", "LAZADA", "ZALO"]
danh_muc = ["Giay", "Dep", "Quan", "Ao"]
array_order = []
for i in range(1, 22):
    don_hang = {}
    # ma don hang viet tat la mdh
    if i < 10:
        ma = "mdh00" + str(i)
    else:
        ma = "mdh0" + str(i)
    don_hang["ma_don"] = ma
    # ngay dat don hang
    thang = i % 12 + 1
    if thang < 10:
        thang = "0" + str(thang)
    else:
        thang = str(thang)
    ngay = i + 3
    if ngay > 28:
        ngay = ngay - 28
    if ngay < 10:
        ngay = "0" + str(ngay)
    else:
        ngay = str(ngay)
    don_hang["ngay_dat"] = "2025-" + thang + "-" + ngay
    # khu vuc
    if i % 4 == 1:
        don_hang["khu_vuc"] = khu_vuc[0]
    elif i % 4 == 2:
        don_hang["khu_vuc"] = khu_vuc[1]
    elif i % 4 == 3:
        don_hang["khu_vuc"] = khu_vuc[2]
    else:
        don_hang["khu_vuc"] = khu_vuc[3]
    # kenh ban
    if i % 4 == 1:
        don_hang["kenh_ban"] = kenh_ban[0]
    elif i % 4 == 2:
        don_hang["kenh_ban"] = kenh_ban[1]
    elif i % 4 == 3:
        don_hang["kenh_ban"] = kenh_ban[2]
    else:
        don_hang["kenh_ban"] = kenh_ban[3]
    # danh muc
    if i % 4 == 1:
        don_hang["danh_muc"] = danh_muc[0]
    elif i % 4 == 2:
        don_hang["danh_muc"] = danh_muc[1]
    elif i % 4 == 3:
        don_hang["danh_muc"] = danh_muc[2]
    else:
        don_hang["danh_muc"] = danh_muc[3]
    # gia tri don hang
    gia = 500000
    gia = gia + i * 100000
    don_hang["gia_tri"] = gia
    # trang thai cua don hang
    if i % 4 == 1:
        don_hang["trang_thai"] = trang_thai_don_hang[0]
    elif i % 4 == 2:
        don_hang["trang_thai"] = trang_thai_don_hang[1]
    elif i % 4 == 3:
        don_hang["trang_thai"] = trang_thai_don_hang[2]
    else:
        don_hang["trang_thai"] = trang_thai_don_hang[3]
    array_order.append(don_hang)
# lay nhung muc khong trung
tap_danh_muc = set()
for don in array_order:
    tap_danh_muc.add(don["danh_muc"])
print("Tong so don hang:", len(array_order))
print("Danh muc khong trung:", tap_danh_muc)
print("\n3 don hang dau:")
for i in range(3):
    print(array_order[i])
```

```python
# Phần 1.2
# 1.2.a
def loc_don_hang(ds, **tieu_chi):
    danh_sach_ket_qua = []
    for don_hang in ds:
        dung_het = True
        for ten_tieu_chi, gia_tri_can_tim in tieu_chi.items():
            if don_hang[ten_tieu_chi] != gia_tri_can_tim:
                dung_het = False
                break
        if dung_het == True:
            danh_sach_ket_qua.append(don_hang)
    return danh_sach_ket_qua
# 1.2.b
def thong_ke_theo_nhom(ds, truong_nhom):
    ket_qua_thong_ke = {}
    for don_hang in ds:
        ten_nhom = don_hang[truong_nhom]
        if ten_nhom not in ket_qua_thong_ke:
            ket_qua_thong_ke[ten_nhom] = {
                "so_don": 0,
                "tong_gia_tri": 0,
                "gia_tri_tb": 0,
                "ty_le_hoan_thanh": 0,
                "so_don_da_giao": 0
            }
        ket_qua_thong_ke[ten_nhom]["so_don"] += 1
        ket_qua_thong_ke[ten_nhom]["tong_gia_tri"] += don_hang["gia_tri"]
        if don_hang["trang_thai"] == "Da giao":
            ket_qua_thong_ke[ten_nhom]["so_don_da_giao"] += 1
    for ten_nhom in ket_qua_thong_ke:
        tong = ket_qua_thong_ke[ten_nhom]["tong_gia_tri"]
        so = ket_qua_thong_ke[ten_nhom]["so_don"]
        ket_qua_thong_ke[ten_nhom]["gia_tri_tb"] = tong / so
        ket_qua_thong_ke[ten_nhom]["ty_le_hoan_thanh"] = (
            ket_qua_thong_ke[ten_nhom]["so_don_da_giao"] * 100
        ) / so
        del ket_qua_thong_ke[ten_nhom]["so_don_da_giao"]
    return ket_qua_thong_ke
#1.2.c
def phan_khuc_don_hang(ds):
    danh_sach_sau_khi_phan_khuc = []
    for don_hang in ds:
        don_hang_moi = {}
        for ten_thuoc_tinh, gia_tri in don_hang.items():
            don_hang_moi[ten_thuoc_tinh] = gia_tri
        if don_hang["gia_tri"] < 1500000:
            don_hang_moi["phan_khuc"] = "Thap"
        elif don_hang["gia_tri"] < 2500000:
            don_hang_moi["phan_khuc"] = "Trung binh"

        else:
            don_hang_moi["phan_khuc"] = "Cao"
        danh_sach_sau_khi_phan_khuc.append(don_hang_moi)
    return danh_sach_sau_khi_phan_khuc
# 1.2.d
def bao_cao_tom_tat(ds):
    tong_doanh_thu = 0
    dem_theo_kenh = {}
    dem_theo_khu_vuc = {}
    for don_hang in ds:
        if don_hang["trang_thai"] == "Da giao":
            tong_doanh_thu += don_hang["gia_tri"]
        ten_kenh = don_hang["kenh_ban"]
        if ten_kenh not in dem_theo_kenh:
            dem_theo_kenh[ten_kenh] = 1
        else:
            dem_theo_kenh[ten_kenh] += 1
        ten_khu_vuc = don_hang["khu_vuc"]
        if ten_khu_vuc not in dem_theo_khu_vuc:
            dem_theo_khu_vuc[ten_khu_vuc] = 1
        else:
            dem_theo_khu_vuc[ten_khu_vuc] += 1
    kenh_nhieu_nhat = ""
    so_don_kenh = 0
    for ten_kenh in dem_theo_kenh:
        if dem_theo_kenh[ten_kenh] > so_don_kenh:
            so_don_kenh = dem_theo_kenh[ten_kenh]
            kenh_nhieu_nhat = ten_kenh
    khu_vuc_nhieu_nhat = ""
    so_don_khu_vuc = 0
    for ten_khu_vuc in dem_theo_khu_vuc:
        if dem_theo_khu_vuc[ten_khu_vuc] > so_don_khu_vuc:
            so_don_khu_vuc = dem_theo_khu_vuc[ten_khu_vuc]
            khu_vuc_nhieu_nhat = ten_khu_vuc
    print("Bao cao")
    print()
    print("Tong so don hang =", len(ds))
    print("Tong doanh thu =", tong_doanh_thu)
    print("Top kenh ban =", kenh_nhieu_nhat)
    print("So don cua kenh =", so_don_kenh)
    print("Top khu vuc =", khu_vuc_nhieu_nhat)
    print("So don cua khu vuc =", so_don_khu_vuc)
```

```python
# Phan 2.1
# class khach hang
class KhachHang:
    def __init__(self, ma_kh, ho_ten, email, dia_chi, ngay_dang_ky):
        self.ma_kh = ma_kh
        self.ho_ten = ho_ten
        self.email = email
        self.dia_chi = dia_chi
        self.ngay_dang_ky = ngay_dang_ky
    # Property email
    @property
    def email(self):
        return self._email
    # Setter email
    @email.setter
    def email(self, value):
        if "@" not in value:
            raise ValueError("Email khong hop le!")
        self._email = value
    # hien thi thong tin
    def __str__(self):
        return ("Ma KH: " + self.ma_kh +" | Ten: " + self.ho_ten +" | Email: " + self.email +" | Dia chi: " + self.dia_chi)
    # Lich su mua hang
    def lich_su_mua_hang(self, ds_don):
        print("Khach hang", self.ho_ten, "da mua", len(ds_don), "don hang.")
# class Khach Hang VIP
class KhachHangVIP(KhachHang):
    def __init__(self,ma_kh,ho_ten,email,dia_chi,ngay_dang_ky,hang_vip,diem_tich_luy):
        super().__init__(ma_kh,ho_ten,email,dia_chi,ngay_dang_ky)
        self.hang_vip = hang_vip
        self.diem_tich_luy = diem_tich_luy
    # Tinh uu dai
    def tinh_uu_dai(self):
        if self.hang_vip == "Silver":
            return 5
        elif self.hang_vip == "Gold":
            return 10
        elif self.hang_vip == "Platinum":
            return 15
        else:
            return 0
    # Ghi de __str__
    def __str__(self):
        return ("Ma KH: " + self.ma_kh +" | Ten: " + self.ho_ten +" | Hang: " + self.hang_vip +" | Diem: " + str(self.diem_tich_luy))
# class San Pham
class SanPham:
    def __init__(self,ma_sp,ten,danh_muc,gia,ton_kho,muc_canh_bao):
        self.ma_sp = ma_sp
        self.ten = ten
        self.danh_muc = danh_muc
        self.gia = gia
        self.ton_kho = ton_kho
        self.muc_canh_bao = muc_canh_bao
    # Kiem tra co can nhap them hay khong
    def can_nhap_hang(self):
        if self.ton_kho <= self.muc_canh_bao:
            return True
        else:
            return False
    # Gia sau thue
    def tinh_gia_sau_thue(self):
        gia_sau_thue = self.gia * 1.1
        return gia_sau_thue
    # repr
    def __repr__(self):
        return ("Ma SP: " + self.ma_sp +" | Ten: " + self.ten +" | Gia: " + str(self.gia))
    # Sap xep theo gia
    def __lt__(self, other):
        if self.gia < other.gia:
            return True
        else:
            return False
# PHAN 2.2
# kiem thu các class phan 2.1
print("TAO DOI TUONG")
kh1 = KhachHang("KH001","Test1","test1@gmail.com","Ha Noi","01/01/2026")
kh2 = KhachHang("KH002","Test2","test2@gmail.com","Ha Noi","02/01/2026")
vip1 = KhachHangVIP("KH003","Test3","test3@gmail.com","Ha Noi","03/01/2026","Gold",250)
sp1 = SanPham("SP001","Ao","Thoi trang",350000,20,5)
sp2 = SanPham("SP002","Laptop","Dien tu",25000000,2,5)
print("Da tao thanh cong 5 doi tuong.")
# a. sorted()
print()
print("SAP XEP SAN PHAM")
ds_sp = []
ds_sp.append(sp1)
ds_sp.append(sp2)
ds_sp = sorted(ds_sp)
for sp in ds_sp:
    print(sp)
# b. isinstance()
print()
print("PHAN LOAI DOI TUONG")
ds = []
ds.append(kh1)
ds.append(kh2)
ds.append(vip1)
ds.append(sp1)
ds.append(sp2)
for obj in ds:
    if isinstance(obj, KhachHangVIP):
        print(obj.ma_kh, "-", obj.ho_ten, ": Khach Hang VIP")
    elif isinstance(obj, KhachHang):
        print(obj.ma_kh, "-", obj.ho_ten, ": Khach Hang")
    elif isinstance(obj, SanPham):
        print(obj.ma_sp, "-", obj.ten, ": San Pham")
# c. Goi cac phuong thuc
print()
print("Test thu cac phuong thuc")
print(kh1)
kh1.lich_su_mua_hang(["DH001","DH002","DH003"])
print()
print(vip1)
print("Uu dai:", vip1.tinh_uu_dai(), "%")
print()
print(sp1)
print("Gia sau thue:", sp1.tinh_gia_sau_thue())
print("Can nhap hang:", sp2.can_nhap_hang())
# d. try except
print()
print("KIEM TRA EMAIL")
try:
    loi = KhachHang("KH999","Test Loi","test1","Ha Noi","20/06/2026")
except Exception as e:
    print("Loi:", e)
```

```python
# Phan 3
# thu vien dung cho phan 3 nay
import numpy as np
import pandas as pd
# them doan nay de in ra khong con hien e+07
pd.set_option("display.float_format", lambda x: "{:,.2f}".format(x))
# Phan 3.1
# Set random seed de ket qua co the tai lap lai
np.random.seed(2026)
so_don_hang = 350
# Tao cac danh sach gia tri mau
ds_kenh = ["Shopee", "Lazada", "Tiki", "Website", "Cua hang"]
ds_trang_thai = ["done","done","done","shipping","cancelled"]
ds_khu_vuc = ["Ha Dong","Cau Giay","Nam Tu Liem","Thanh Tri"]
ds_danh_muc = ["Thoi trang","Dien tu","Gia dung","Me va Be"]
ma_don = []
for i in range(so_don_hang):
    ma = "DH"
    if i + 1 < 10:
        ma = ma + "000" + str(i + 1)
    elif i + 1 < 100:
        ma = ma + "00" + str(i + 1)
    else:
        ma = ma + "0" + str(i + 1)
    ma_don.append(ma)
gia_tri = np.random.randint(100000,2000000,so_don_hang)
chiet_khau = np.random.choice([0,0.05,0.1,0.15,0.2],so_don_hang)
so_luong = np.random.randint(1, 10,so_don_hang)
kenh_ban = np.random.choice(ds_kenh,so_don_hang)
trang_thai = np.random.choice(ds_trang_thai,so_don_hang)
khu_vuc = np.random.choice(ds_khu_vuc,so_don_hang)
danh_muc = np.random.choice(ds_danh_muc,so_don_hang)
ngay_dat = pd.date_range(start="2026-01-01", periods=so_don_hang)
# Tao DataFrame
df_orders = pd.DataFrame()
df_orders["ma_don"] = ma_don
df_orders["gia_tri"] = gia_tri
df_orders["chiet_khau"] = chiet_khau
df_orders["so_luong"] = so_luong
df_orders["kenh_ban"] = kenh_ban
df_orders["trang_thai"] = trang_thai
df_orders["khu_vuc"] = khu_vuc
df_orders["danh_muc"] = danh_muc
df_orders["ngay"] = ngay_dat
# Tao mot vai loi du lieu de lam sach
# Tao gia tri rong
df_orders.loc[5,"gia_tri"] = np.nan
df_orders.loc[10,"gia_tri"] = np.nan
df_orders.loc[15,"gia_tri"] = np.nan
df_orders.loc[20,"chiet_khau"] = np.nan
df_orders.loc[21,"chiet_khau"] = np.nan
# Tao trung lap
df_orders.loc[350] = df_orders.loc[0]
df_orders.loc[351] = df_orders.loc[1]
# Tao outlier
df_orders.loc[50,"gia_tri"] = 50000000
# Bao cao truoc khi lam sach
print("data truoc khi lam sach la:")
print()
print("Tong so dong:")
print(len(df_orders))
print()
print("So gia tri rong:")
print(df_orders.isnull().sum())
print()
print("So dong trung lap:")
print(df_orders.duplicated().sum())
print()
# Bat dau lam sach
# Xoa trung lap
df_clean = df_orders.drop_duplicates().copy()
# Dien gia tri thieu
gia_tri_trung_vi = df_clean["gia_tri"].median()
df_clean["gia_tri"] = df_clean["gia_tri"].fillna(gia_tri_trung_vi)
df_clean["chiet_khau"] = df_clean["chiet_khau"].fillna(0)
# Tim outlier bang IQR
Q1 = df_clean["gia_tri"].quantile(0.25)
Q3 = df_clean["gia_tri"].quantile(0.75)
IQR = Q3 - Q1
gioi_han_duoi = Q1 - IQR * 1.5
gioi_han_tren = Q3 + IQR * 1.5
# Thay outlier bang gioi han tren
df_clean.loc[df_clean["gia_tri"] > gioi_han_tren,"gia_tri"] = gioi_han_tren
# Bao cao sau khi lam sach
print("data sau khi lam sach la:")
print()
print("Tong so dong:")
print(len(df_clean))
print()
print("Tong so gia tri rong:")
print(df_clean.isnull().sum().sum())
print()
print("Tong so dong trung lap:")
print(df_clean.duplicated().sum())
print()
print("5 dong dau tien:")
print(df_clean.head())
# Luu file CSV
df_clean.to_csv("du_lieu_don_hang.csv",index=False,encoding="utf-8-sig")
print()
print("Da luu file du_lieu_don_hang.csv")
```

```python
# PHAN 4.1
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import StrMethodFormatter
print("Dang ve dashboard...")
# Tao Figure
fig = plt.figure(figsize=(14,10))
# Chia man hinh thanh 2 hang 2 cot
grid = fig.add_gridspec(2,2)
# BIEU DO 1 - XU HUONG DOANH THU THEO THANG
ax1 = fig.add_subplot(grid[0,0])
# Lay cac don hang da giao
bang_doanh_thu = df_clean[df_clean["trang_thai"] == "done"].copy()
# Tao cot thang
bang_doanh_thu["thang"] = bang_doanh_thu["ngay"].dt.to_period("M")
# Tong doanh thu tung thang
tong_theo_thang = bang_doanh_thu.groupby("thang")["gia_tri"].sum()
# Rolling Average 3 thang
rolling = tong_theo_thang.rolling(3).mean()
# Tao truc x
x = range(len(tong_theo_thang))
# Ve doanh thu
ax1.plot(x,tong_theo_thang.values,color="blue",linewidth=2,marker="o",label="Doanh thu")
# Ve Rolling Average
ax1.plot(x,rolling.values,color="red",linewidth=2,linestyle="--",marker="s",label="Rolling Average")
# To mau
ax1.fill_between(x,tong_theo_thang.values,color="lightblue",alpha=0.4)
# Tim diem cao nhat
vi_tri_max = np.argmax(tong_theo_thang.values)
ax1.annotate("Cao nhat",xy=(vi_tri_max, tong_theo_thang.values[vi_tri_max]),xytext=(vi_tri_max,tong_theo_thang.values[vi_tri_max] + 5000000),ha="center",arrowprops=dict(arrowstyle="->",color="black"))
# Ten thang
nhan_thang = []
for thang in tong_theo_thang.index:
    nhan_thang.append(thang.strftime("%m"))
ax1.set_xticks(list(x))
ax1.set_xticklabels(nhan_thang)
ax1.set_title("Doanh thu theo thang")
ax1.set_xlabel("Thang")
ax1.set_ylabel("Doanh thu")
# Khong hien 1e7
ax1.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
ax1.legend()
# BIEU DO 2 - BAR CHART
ax2 = fig.add_subplot(grid[0,1])
tong_kenh = df_clean.groupby("kenh_ban")["gia_tri"].sum()
ax2.bar(tong_kenh.index,tong_kenh.values,color="green")
ax2.set_title("Tong doanh thu theo kenh")
ax2.set_xlabel("Kenh ban")
ax2.set_ylabel("Doanh thu")
ax2.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
# BIEU DO 3 - PIE CHART
ax3 = fig.add_subplot(grid[1,0])
dem_trang_thai = df_clean["trang_thai"].value_counts()
ax3.pie(dem_trang_thai.values,labels=dem_trang_thai.index,autopct="%1.1f%%",colors=["gold","orange","skyblue"])
ax3.set_title("Ty le trang thai don hang")
# BIEU DO 4 - SCATTER
ax4 = fig.add_subplot(grid[1,1])
x = df_clean["so_luong"]
y = df_clean["gia_tri"]
ax4.scatter(x,y,color="purple")
# Ve duong xu huong
he_so = np.polyfit(x,y,1)
duong = np.poly1d(he_so)
ax4.plot(x,duong(x),color="red",linewidth=2,label="Trend Line")
ax4.set_title("Gia tri va so luong")
ax4.set_xlabel("So luong")
ax4.set_ylabel("Gia tri")
ax4.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
ax4.legend()
# TIEU DE CHUNG
fig.suptitle("Dashboard Phan Tich Don Hang",fontsize=16)
plt.tight_layout()
plt.savefig(
    "dashboard_matplotlib.png",
    dpi=200
)
plt.show()
print()
print("Da luu file dashboard_matplotlib.png")
```
