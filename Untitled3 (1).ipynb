```python
# PHẦN 2: LẬP TRÌNH HƯỚNG ĐỐI TƯỢNG 
# 2.1. Thiết kế phân cấp lớp cho hệ thống 

class KhachHang:
    """Lớp khách hàng cơ bản."""
    def __init__(self, ma_kh, ho_ten, email, dia_chi, ngay_dang_ky):
        self.ma_kh = ma_kh
        self.ho_ten = ho_ten
        self.email = email
        self.dia_chi = dia_chi
        self.ngay_dang_ky = ngay_dang_ky
    @property
    def email(self):
        return self._email
    @email.setter
    def email(self, gia_tri):
        """Kiểm tra tính hợp lệ của định dạng email (phải chứa '@' và dấu chấm ở phần tên miền)."""
        if '@' not in gia_tri or '.' not in gia_tri.split('@')[-1]:
            raise ValueError('Email không đúng định dạng.')
        self._email = gia_tri
    def lich_su_mua_hang(self, ds_don):
        """Lọc và trả về danh sách lịch sử đơn hàng của khách hàng dựa vào mã khách hàng."""
        return [don for don in ds_don if don.get('ma_kh') == self.ma_kh]
    def __str__(self):
       return f'{self.ma_kh} - {self.ho_ten} - {self.email}'
    def __lt__(self, other):
        """Sắp xếp khách hàng theo ngày đăng ký."""
        return self.ngay_dang_ky < other.ngay_dang_ky

class KhachHangVIP(KhachHang):
    """Lớp khách hàng VIP kế thừa từ KhachHang dựa vào mã khách hàng."""
    def __init__(self, ma_kh, ho_ten, email, dia_chi, ngay_dang_ky,
                 hang_vip, diem_tich_luy):
        super().__init__(ma_kh, ho_ten, email, dia_chi, ngay_dang_ky)
        self.hang_vip = hang_vip
        self.diem_tich_luy = diem_tich_luy
    def tinh_uu_dai(self):
        muc_uu_dai = {
            'Silver': 0.05,
            'Gold': 0.10,
            'Platinum': 0.15
        }
        return muc_uu_dai.get(self.hang_vip, 0)
    def __str__(self):
        """Ghi đè phương thức __str__ để hiển thị thêm thông tin chi tiết VIP."""
        return (f'{super().__str__()} - VIP {self.hang_vip} - '
                f'{self.diem_tich_luy} điểm')

class SanPham:
    """Lớp quản lý thông tin sản phẩm dựa trên dữ liệu sheet Products."""
    def __init__(self, ma_sp, ten, danh_muc, gia, ton_kho, muc_canh_bao):
        self.ma_sp = ma_sp
        self.ten = ten
        self.danh_muc = danh_muc
        self.gia = gia
        self.ton_kho = ton_kho
        self.muc_canh_bao = muc_canh_bao
    def can_nhap_hang(self):
        return self.ton_kho <= self.muc_canh_bao
    def tinh_gia_sau_thue(self, thue_suat=0.10):
        return self.gia * (1 + thue_suat)
    def __repr__(self):
        return f"SanPham('{self.ma_sp}', '{self.ten}', ton_kho={self.ton_kho})"
```

```python
# 2.2. Tích hợp và kiểm thử hệ thống lớp 
# Tạo ít nhất 5 đối tượng từ các lớp đã định nghĩa
kh1 = KhachHang('KH004', 'Ngô Thị Bình', 'kh004@gmail.com', 'Hà Nội', '2024-02-05')
kh2 = KhachHang('KH005', 'Lê Thu Nhân', 'kh005@gmail.com', 'TP.HCM', '2024-07-29')
kh3 = KhachHangVIP('KH001', 'Phạm Văn Sơn', 'kh001@gmail.com', 'Hải Phòng', '2024-11-14', 'Gold', 2111)
kh4 = KhachHangVIP('KH002', 'Vũ Thanh Hòa', 'kh002@gmail.com', 'Hà Nội', '2024-05-19', 'Platinum', 3372)
kh5 = KhachHangVIP('KH003', 'Phạm Phương Hà', 'kh003@gmail.com', 'Đà Nẵng', '2024-12-07', 'Platinum', 5982)
sp1 = SanPham('SP001', 'Laptop 1', 'Laptop', 26300000, 38, 20)
sp2 = SanPham('SP002', 'Tai nghe 2', 'Audio', 17500000, 12, 20)    

# (a) Sắp xếp list đối tượng theo thuộc tính bằng sorted() tận dụng __lt__
ds_khach = [kh1, kh2, kh3, kh4, kh5]
print('(a) SẮP XẾP DANH SÁCH KHÁCH HÀNG THEO NGÀY ĐĂNG KÝ:')
for kh in sorted(ds_khach):
    print(kh)
# (b) Dùng isinstance() để phân loại đối tượng
print('\n(b) PHÂN LOẠI KHÁCH HÀNG:')
for kh in ds_khach:
    loai = 'Khach VIP' if isinstance(kh, KhachHangVIP) else 'Khách thường'
    print(kh.ma_kh, '-', loai)
# (c) Gọi đủ tất cả các phương thức đã định nghĩa và kiểm tra kết quả
print('\n(c) MỨC ƯU ĐÃI CỦA KHÁCH VIP:')
for kh in ds_khach:
    if isinstance(kh, KhachHangVIP):
        print(kh.ho_ten, f'{kh.tinh_uu_dai() * 100:.0f}%')
# Tạo danh sách đơn hàng nhỏ (dữ liệu mẫu) để kiểm tra phương thức lich_su_mua_hang()
ds_don_khach = [
    {'ma_don': 'ORD0001', 'ma_kh': 'KH042', 'gia_tri': 8800000},
    {'ma_don': 'ORD0002', 'ma_kh': 'KH026', 'gia_tri': 38420000},
    {'ma_don': 'ORD0010', 'ma_kh': 'KH004', 'gia_tri': 15000000}
]
print('\nLịch sử mua hàng của KH004 (Ngô Thị Bình):')
print(kh1.lich_su_mua_hang(ds_don_khach))
print('\nKIỂM TRA SẢN PHẨM:')
for sp in [sp1, sp2]:
    print(sp)
    print('Cần nhập hàng:', sp.can_nhap_hang())
    print('Giá sau thuế:', f'{sp.tinh_gia_sau_thue():,.0f} VNĐ')
# (d) Xử lý ít nhất 1 trường hợp lỗi bằng try-except 
print('\n(d) KIỂM TRA GÁN EMAIL SAI:')
try: 
    kh1.email = 'email_khong_hop_le'
except ValueError as loi: 
    print('Đã bắt lỗi:', loi)
    
```
