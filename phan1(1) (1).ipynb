```python
# Phần 1.1
trang_thai_don_hang = ("Da giao", "Dang giao", "Huy don", "Da hoan tra")
khu_vuc = ["HN", "HN1", "HN2", "HN3"]
kenh_ban = ["TIKTOK", "SHOPPE", "LAZADA", "ZALO"]
danh_muc = ["Giay", "Dep", "Quan", "Ao"]
array_order = []
for i in range(1, 22):
    don_hang = {}
    # ma don hang viet tat la mdh
    if i < 10:
        ma = "mdh00" + str(i)
    else:
        ma = "mdh0" + str(i)
    don_hang["ma_don"] = ma
    # ngay dat don hang
    thang = i % 12 + 1
    if thang < 10:
        thang = "0" + str(thang)
    else:
        thang = str(thang)
    ngay = i + 3
    if ngay > 28:
        ngay = ngay - 28
    if ngay < 10:
        ngay = "0" + str(ngay)
    else:
        ngay = str(ngay)
    don_hang["ngay_dat"] = "2025-" + thang + "-" + ngay
    # khu vuc
    if i % 4 == 1:
        don_hang["khu_vuc"] = khu_vuc[0]
    elif i % 4 == 2:
        don_hang["khu_vuc"] = khu_vuc[1]
    elif i % 4 == 3:
        don_hang["khu_vuc"] = khu_vuc[2]
    else:
        don_hang["khu_vuc"] = khu_vuc[3]
    # kenh ban
    if i % 4 == 1:
        don_hang["kenh_ban"] = kenh_ban[0]
    elif i % 4 == 2:
        don_hang["kenh_ban"] = kenh_ban[1]
    elif i % 4 == 3:
        don_hang["kenh_ban"] = kenh_ban[2]
    else:
        don_hang["kenh_ban"] = kenh_ban[3]
    # danh muc
    if i % 4 == 1:
        don_hang["danh_muc"] = danh_muc[0]
    elif i % 4 == 2:
        don_hang["danh_muc"] = danh_muc[1]
    elif i % 4 == 3:
        don_hang["danh_muc"] = danh_muc[2]
    else:
        don_hang["danh_muc"] = danh_muc[3]
    # gia tri don hang
    gia = 500000
    gia = gia + i * 100000
    don_hang["gia_tri"] = gia
    # trang thai cua don hang
    if i % 4 == 1:
        don_hang["trang_thai"] = trang_thai_don_hang[0]
    elif i % 4 == 2:
        don_hang["trang_thai"] = trang_thai_don_hang[1]
    elif i % 4 == 3:
        don_hang["trang_thai"] = trang_thai_don_hang[2]
    else:
        don_hang["trang_thai"] = trang_thai_don_hang[3]
    array_order.append(don_hang)
# lay nhung muc khong trung
tap_danh_muc = set()
for don in array_order:
    tap_danh_muc.add(don["danh_muc"])
print("Tong so don hang:", len(array_order))
print("Danh muc khong trung:", tap_danh_muc)
print("\n3 don hang dau:")
for i in range(3):
    print(array_order[i])
```

```python
# Phần 1.2
# 1.2.a
def loc_don_hang(ds, **tieu_chi):
    danh_sach_ket_qua = []
    for don_hang in ds:
        dung_het = True
        for ten_tieu_chi, gia_tri_can_tim in tieu_chi.items():
            if don_hang[ten_tieu_chi] != gia_tri_can_tim:
                dung_het = False
                break
        if dung_het == True:
            danh_sach_ket_qua.append(don_hang)
    return danh_sach_ket_qua
# 1.2.b
def thong_ke_theo_nhom(ds, truong_nhom):
    ket_qua_thong_ke = {}
    for don_hang in ds:
        ten_nhom = don_hang[truong_nhom]
        if ten_nhom not in ket_qua_thong_ke:
            ket_qua_thong_ke[ten_nhom] = {
                "so_don": 0,
                "tong_gia_tri": 0,
                "gia_tri_tb": 0,
                "ty_le_hoan_thanh": 0,
                "so_don_da_giao": 0
            }
        ket_qua_thong_ke[ten_nhom]["so_don"] += 1
        ket_qua_thong_ke[ten_nhom]["tong_gia_tri"] += don_hang["gia_tri"]
        if don_hang["trang_thai"] == "Da giao":
            ket_qua_thong_ke[ten_nhom]["so_don_da_giao"] += 1
    for ten_nhom in ket_qua_thong_ke:
        tong = ket_qua_thong_ke[ten_nhom]["tong_gia_tri"]
        so = ket_qua_thong_ke[ten_nhom]["so_don"]
        ket_qua_thong_ke[ten_nhom]["gia_tri_tb"] = tong / so
        ket_qua_thong_ke[ten_nhom]["ty_le_hoan_thanh"] = (
            ket_qua_thong_ke[ten_nhom]["so_don_da_giao"] * 100
        ) / so
        del ket_qua_thong_ke[ten_nhom]["so_don_da_giao"]
    return ket_qua_thong_ke
#1.2.c
def phan_khuc_don_hang(ds):
    danh_sach_sau_khi_phan_khuc = []
    for don_hang in ds:
        don_hang_moi = {}
        for ten_thuoc_tinh, gia_tri in don_hang.items():
            don_hang_moi[ten_thuoc_tinh] = gia_tri
        if don_hang["gia_tri"] < 1500000:
            don_hang_moi["phan_khuc"] = "Thap"
        elif don_hang["gia_tri"] < 2500000:
            don_hang_moi["phan_khuc"] = "Trung binh"

        else:
            don_hang_moi["phan_khuc"] = "Cao"
        danh_sach_sau_khi_phan_khuc.append(don_hang_moi)
    return danh_sach_sau_khi_phan_khuc
# 1.2.d
def bao_cao_tom_tat(ds):
    tong_doanh_thu = 0
    dem_theo_kenh = {}
    dem_theo_khu_vuc = {}
    for don_hang in ds:
        if don_hang["trang_thai"] == "Da giao":
            tong_doanh_thu += don_hang["gia_tri"]
        ten_kenh = don_hang["kenh_ban"]
        if ten_kenh not in dem_theo_kenh:
            dem_theo_kenh[ten_kenh] = 1
        else:
            dem_theo_kenh[ten_kenh] += 1
        ten_khu_vuc = don_hang["khu_vuc"]
        if ten_khu_vuc not in dem_theo_khu_vuc:
            dem_theo_khu_vuc[ten_khu_vuc] = 1
        else:
            dem_theo_khu_vuc[ten_khu_vuc] += 1
    kenh_nhieu_nhat = ""
    so_don_kenh = 0
    for ten_kenh in dem_theo_kenh:
        if dem_theo_kenh[ten_kenh] > so_don_kenh:
            so_don_kenh = dem_theo_kenh[ten_kenh]
            kenh_nhieu_nhat = ten_kenh
    khu_vuc_nhieu_nhat = ""
    so_don_khu_vuc = 0
    for ten_khu_vuc in dem_theo_khu_vuc:
        if dem_theo_khu_vuc[ten_khu_vuc] > so_don_khu_vuc:
            so_don_khu_vuc = dem_theo_khu_vuc[ten_khu_vuc]
            khu_vuc_nhieu_nhat = ten_khu_vuc
    print("Bao cao")
    print()
    print("Tong so don hang =", len(ds))
    print("Tong doanh thu =", tong_doanh_thu)
    print("Top kenh ban =", kenh_nhieu_nhat)
    print("So don cua kenh =", so_don_kenh)
    print("Top khu vuc =", khu_vuc_nhieu_nhat)
    print("So don cua khu vuc =", so_don_khu_vuc)
```

```python
# ham de test cac ham tren
print("Kiem tra ham loc_don_hang")
print(loc_don_hang(array_order, khu_vuc="HN"))
print()
print("Kiem tra ham thong_ke_theo_nhom")
print(thong_ke_theo_nhom(array_order, "khu_vuc"))
print()
print("Kiem tra ham phan_khuc_don_hang")
ds_moi = phan_khuc_don_hang(array_order)
for i in range(3):
    print(ds_moi[i])
print()
print("Kiem tra ham bao_cao_tom_tat")
bao_cao_tom_tat(array_order)
```
