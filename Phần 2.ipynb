```python
# Phan 2.1
# class khach hang
class KhachHang:
    def __init__(self, ma_kh, ho_ten, email, dia_chi, ngay_dang_ky):
        self.ma_kh = ma_kh
        self.ho_ten = ho_ten
        self.email = email
        self.dia_chi = dia_chi
        self.ngay_dang_ky = ngay_dang_ky
    # Property email
    @property
    def email(self):
        return self._email
    # Setter email
    @email.setter
    def email(self, value):
        if "@" not in value:
            raise ValueError("Email khong hop le!")
        self._email = value
    # hien thi thong tin
    def __str__(self):
        return ("Ma KH: " + self.ma_kh +" | Ten: " + self.ho_ten +" | Email: " + self.email +" | Dia chi: " + self.dia_chi)
    # Lich su mua hang
    def lich_su_mua_hang(self, ds_don):
        print("Khach hang", self.ho_ten, "da mua", len(ds_don), "don hang.")
# class Khach Hang VIP
class KhachHangVIP(KhachHang):
    def __init__(self,ma_kh,ho_ten,email,dia_chi,ngay_dang_ky,hang_vip,diem_tich_luy):
        super().__init__(ma_kh,ho_ten,email,dia_chi,ngay_dang_ky)
        self.hang_vip = hang_vip
        self.diem_tich_luy = diem_tich_luy
    # Tinh uu dai
    def tinh_uu_dai(self):
        if self.hang_vip == "Silver":
            return 5
        elif self.hang_vip == "Gold":
            return 10
        elif self.hang_vip == "Platinum":
            return 15
        else:
            return 0
    # Ghi de __str__
    def __str__(self):
        return ("Ma KH: " + self.ma_kh +" | Ten: " + self.ho_ten +" | Hang: " + self.hang_vip +" | Diem: " + str(self.diem_tich_luy))
# class San Pham
class SanPham:
    def __init__(self,ma_sp,ten,danh_muc,gia,ton_kho,muc_canh_bao):
        self.ma_sp = ma_sp
        self.ten = ten
        self.danh_muc = danh_muc
        self.gia = gia
        self.ton_kho = ton_kho
        self.muc_canh_bao = muc_canh_bao
    # Kiem tra co can nhap them hay khong
    def can_nhap_hang(self):
        if self.ton_kho <= self.muc_canh_bao:
            return True
        else:
            return False
    # Gia sau thue
    def tinh_gia_sau_thue(self):
        gia_sau_thue = self.gia * 1.1
        return gia_sau_thue
    # repr
    def __repr__(self):
        return ("Ma SP: " + self.ma_sp +" | Ten: " + self.ten +" | Gia: " + str(self.gia))
    # Sap xep theo gia
    def __lt__(self, other):
        if self.gia < other.gia:
            return True
        else:
            return False
# PHAN 2.2
# kiem thu các class phan 2.1
print("TAO DOI TUONG")
kh1 = KhachHang("KH001","Test1","test1@gmail.com","Ha Noi","01/01/2026")
kh2 = KhachHang("KH002","Test2","test2@gmail.com","Ha Noi","02/01/2026")
vip1 = KhachHangVIP("KH003","Test3","test3@gmail.com","Ha Noi","03/01/2026","Gold",250)
sp1 = SanPham("SP001","Ao","Thoi trang",350000,20,5)
sp2 = SanPham("SP002","Laptop","Dien tu",25000000,2,5)
print("Da tao thanh cong 5 doi tuong.")
# a. sorted()
print()
print("SAP XEP SAN PHAM")
ds_sp = []
ds_sp.append(sp1)
ds_sp.append(sp2)
ds_sp = sorted(ds_sp)
for sp in ds_sp:
    print(sp)
# b. isinstance()
print()
print("PHAN LOAI DOI TUONG")
ds = []
ds.append(kh1)
ds.append(kh2)
ds.append(vip1)
ds.append(sp1)
ds.append(sp2)
for obj in ds:
    if isinstance(obj, KhachHangVIP):
        print(obj.ma_kh, "-", obj.ho_ten, ": Khach Hang VIP")
    elif isinstance(obj, KhachHang):
        print(obj.ma_kh, "-", obj.ho_ten, ": Khach Hang")
    elif isinstance(obj, SanPham):
        print(obj.ma_sp, "-", obj.ten, ": San Pham")
# c. Goi cac phuong thuc
print()
print("Test thu cac phuong thuc")
print(kh1)
kh1.lich_su_mua_hang(["DH001","DH002","DH003"])
print()
print(vip1)
print("Uu dai:", vip1.tinh_uu_dai(), "%")
print()
print(sp1)
print("Gia sau thue:", sp1.tinh_gia_sau_thue())
print("Can nhap hang:", sp2.can_nhap_hang())
# d. try except
print()
print("KIEM TRA EMAIL")
try:
    loi = KhachHang("KH999","Test Loi","test1","Ha Noi","20/06/2026")
except Exception as e:
    print("Loi:", e)
```
