```python
# Phần 1.1

trang_thai_don_hang = ('Da giao', 'Dang giao', 'Huy don', 'Da hoan tra')
khu_vuc_mau = ['HN', 'HN1', 'HN2', 'HN3']
kenh_ban_mau = ['a', 'b', 'c', 'd']
danh_muc = ['d', 'e', 'f', 'g']

array_order = []

# Tạo 24 đơn hàng
for i in range(1, 25):
    order = {
        'ma_don': f'DH{i:03d}',
        'ngay_dat': f'2025-{((i + 2) % 12) + 1:02d}-{((i * 4) % 28) + 1:02d}',
        'khu_vuc': khu_vuc_mau[(i + 1) % len(khu_vuc_mau)],
        'kenh_ban': kenh_ban_mau[(i * 3) % len(kenh_ban_mau)],
        'danh_muc': danh_muc[(i * 2 + 1) % len(danh_muc)],
        'gia_tri': 500000 + i * 100000,
        'trang_thai': trang_thai_don_hang[(i + 2) % len(trang_thai_don_hang)]
    }

    array_order.append(order)

# Lấy danh sách danh mục không trùng
tap_danh_muc = set()

for order in array_order:
    tap_danh_muc.add(order['danh_muc'])

print("Tong so don hang:", len(array_order))
print("Danh muc khong trung:", tap_danh_muc)

print("\n3 don hang dau:")
for order in array_order[:3]:
    print(order)
```
